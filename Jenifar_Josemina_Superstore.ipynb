**TASK1-Convert Order Date to Date time & extract Year,Month and Day Name**

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Day_Name'] = df['Order Date'].dt.day_name()

display(df.head())
```

```python
df['Category_Total_Sales'] = df.groupby('Category')['Sales'].transform('sum')
df['Percentage_Contribution'] = (df['Sales'] / df['Category_Total_Sales']) * 100

display(df[['Category', 'Sales', 'Category_Total_Sales', 'Percentage_Contribution']].head())
```

**TASK2**- Matplotlib 3D surface plot of Synthetic Sales-Profit Surface

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Get the range of Sales and Profit from the DataFrame
sales_min, sales_max = df['Sales'].min(), df['Sales'].max()
profit_min, profit_max = df['Profit'].min(), df['Profit'].max()

# Create a meshgrid for the X (Sales) and Y (Profit) axes
X = np.linspace(sales_min, sales_max, 100)
Y = np.linspace(profit_min, profit_max, 100)
X, Y = np.meshgrid(X, Y)

# Define a synthetic Z-axis function (e.g., based on Sales and Profit)
# This is a hypothetical function for demonstration purposes
# For example, Z could represent an idealized 'satisfaction score' or 'growth potential'
Z = np.sin(X/sales_max * np.pi * 2) + np.cos(Y/profit_max * np.pi * 2) # Example synthetic function
Z = (X/sales_max)**2 + (Y/profit_max)**2 # Another example: a parabolic surface
Z = np.sqrt(X/sales_max) * np.log(np.maximum(Y, 1)/profit_max + 1) * 100 # Example more complex function

# Let's simplify Z for a clearer visualization, perhaps a linear combination scaled
# We'll use a simpler representation for clarity in this example
# Ensure Z doesn't have NaN or inf values if using complex functions

# For a sales-profit surface, let's create a simpler, more intuitive synthetic Z
# For instance, Z could represent some 'performance' metric
Z = (X / sales_max * 10) + (Y / profit_max * 5)

# Create a 3D figure and axes
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surface = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

# Set labels and title
ax.set_xlabel('Sales')
ax.set_ylabel('Profit')
ax.set_zlabel('Synthetic Performance Metric (Z)')
ax.set_title('3D Surface Plot: Synthetic Sales-Profit Surface')

# Add a color bar for Z values
fig.colorbar(surface, shrink=0.5, aspect=10, label='Synthetic Performance Metric')

plt.tight_layout()
plt.show()
```

**TASK10**- Customized dark Theme Animated slider chart

```python
import plotly.graph_objects as go
import numpy as np

# Create figures
fig = go.Figure()

# Get unique discount values and sort them
discount_values = np.sort(df['Discount'].unique())

# Add traces for each discount level
for discount in discount_values:
    df_filtered = df[df['Discount'] == discount]
    fig.add_trace(go.Scatter(
        x=df_filtered['Sales'],
        y=df_filtered['Profit'],
        mode='markers',
        name=f'Discount: {discount:.2f}',
        marker=dict(size=df_filtered['Quantity'] * 2, sizemode='area', sizeref=2.*max(df_filtered['Quantity'])/(40.**2), sizemin=4),
        text=df_filtered['Customer Name'] + '<br>Sales: ' + df_filtered['Sales'].astype(str) + '<br>Profit: ' + df_filtered['Profit'].astype(str),
        visible=False # All traces initially invisible
    ))

# Set the first trace visible
fig.data[0].visible = True

# Create slider steps
steps = []
for i, discount in enumerate(discount_values):
    step = dict(
        method='update',
        args=[{"visible": [False] * len(fig.data)}], # Toggle all traces off
        label=f'{discount:.2f}'
    )
    step['args'][0]['visible'][i] = True # Toggle i'th trace on
    steps.append(step)

sliders = [dict(
    active=0,
    currentvalue={"prefix": "Discount: "},
    pad={"t": 50},
    steps=steps
)]

# Update layout for dark theme and slider
fig.update_layout(
    title_text='Sales vs. Profit by Discount (Interactive Slider)',
    xaxis_title='Sales',
    yaxis_title='Profit',
    height=700,
    sliders=sliders,
    template='plotly_dark' # Apply dark theme
)

fig.show()
```

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px

# Initialize the Dash app for the Discount Histogram. Using a distinct app variable and port.
app_discount_histogram = Dash(__name__)

# Define the app layout
app_discount_histogram.layout = html.Div([
    html.H1("Dynamic Discount Histogram"),
    html.Div([
        html.Label("Select Number of Bins:"),
        dcc.Slider(
            id='bin-slider',
            min=1,
            max=50,
            step=1,
            value=20, # Default number of bins
            marks={i: str(i) for i in range(0, 51, 5)}
        )
    ]),
    dcc.Graph(id='discount-histogram-chart')
])

# Define the callback to update the histogram based on slider selection
@app_discount_histogram.callback(
    Output('discount-histogram-chart', 'figure'),
    Input('bin-slider', 'value')
)
def update_discount_histogram(n_bins):
    fig = px.histogram(df, x='Discount', nbins=n_bins,
                     title=f'Distribution of Discount (with {n_bins} bins)',
                     labels={'Discount': 'Discount Amount', 'count': 'Number of Orders'},
                     marginal='rug', # Add a rug plot to show individual data points
                     template='plotly_white')

    fig.update_layout(bargap=0.1) # Add some gap between bars for better visualization
    return fig

print("Dash app for Dynamic Discount Histogram defined. Execute the next cell to run it.")
```

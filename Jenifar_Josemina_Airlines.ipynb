```python
import pandas as pd

# Load the dataset
df = pd.read_csv("airlines_flights_data.csv")

# Group by airline and calculate total price
airline_sales = df.groupby("airline")["price"].sum().reset_index()

# Rename the column
airline_sales.rename(columns={"price": "Total Sales"}, inplace=True)

# Calculate each airline's percentage contribution
airline_sales["Percentage Contribution"] = (
    airline_sales["Total Sales"] /
    airline_sales["Total Sales"].sum()
) * 100

# Display the result
print(airline_sales)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("airlines_flights_data.csv")

# Create category sales summary
category_sales = df.pivot_table(
    values="price",
    index="airline",
    columns="class",
    aggfunc="sum",
    fill_value=0
).reset_index()

print("Pivoted Data:")
print(category_sales)

# Unpivot using melt()
melted_data = category_sales.melt(
    id_vars="airline",
    var_name="class",
    value_name="total_sales"
)

print("\nUnpivoted Data:")
print(melted_data)
```

```python
import pandas as pd

# Load main dataset
df = pd.read_csv("airlines_flights_data.csv")

# Calculate actual sales for each source city
regional_sales = (
    df.groupby("source_city")["price"]
    .sum()
    .reset_index()
)

regional_sales.rename(
    columns={"price": "actual_sales"},
    inplace=True
)

# Create target sales DataFrame
target_df = regional_sales.copy()

# Set target as 10% higher than actual sales
target_df["target_sales"] = target_df["actual_sales"] * 1.10

# Keep only required columns
target_df = target_df[["source_city", "target_sales"]]

print("Regional Target DataFrame:")
print(target_df)

# Merge target data with main dataset
merged_df = pd.merge(
    df,
    target_df,
    on="source_city",
    how="left"
)

print("\nMerged Data:")
print(merged_df.head())
```

```python
import matplotlib.pyplot as plt

# Group by 'days_left' and sum 'price' to get 'daily sales'
# We'll treat 'days_left' as a proxy for days, sorting it to represent a time series.
daily_sales = df.groupby('days_left')['price'].sum().reset_index()

# Ensure the data is sorted by 'days_left' for correct rolling mean calculation
daily_sales = daily_sales.sort_values(by='days_left')

# Calculate the 7-day rolling mean of 'price'
daily_sales['rolling_mean_7_day'] = daily_sales['price'].rolling(window=7).mean()

# Plot the original daily sales and the 7-day rolling mean
plt.figure(figsize=(14, 7))
plt.plot(daily_sales['days_left'], daily_sales['price'], label='Daily Sales (Aggregated by Days Left)', alpha=0.7)
plt.plot(daily_sales['days_left'], daily_sales['rolling_mean_7_day'], label='7-Day Rolling Mean', color='red')

plt.title('Daily Sales and 7-Day Rolling Mean (based on Days Left)')
plt.xlabel('Days Left Until Flight')
plt.ylabel('Total Sales (Price)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

**TASK3**

```python
import pandas as pd

# Ensure daily_sales is sorted by 'days_left' before applying shift
# This was already done in the previous step, but re-confirming for correctness.
daily_sales = daily_sales.sort_values(by='days_left')

# Create a lag-1 column 'Sales_Prev_Day'
daily_sales['Sales_Prev_Day'] = daily_sales['price'].shift(1)

# Calculate daily percentage change in sales
daily_sales['Daily_Sales_Percentage_Change'] = \
    ((daily_sales['price'] - daily_sales['Sales_Prev_Day']) / daily_sales['Sales_Prev_Day']) * 100

print("Daily Sales with Previous Day's Sales and Percentage Change:")
print(daily_sales.head(10))
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('airlines_flights_data.csv')

# Generate synthetic meshgrid data
x_vals = np.linspace(df['duration'].min(), df['duration'].max(), 50)
y_vals = np.linspace(df['price'].min(), df['price'].max(), 50)
X, Y = np.meshgrid(x_vals, y_vals)
Z = np.sin(np.sqrt(np.abs(X**2 + Y**2)))

# 3D Surface Plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='coolwarm', edgecolor='none', alpha=0.8)

ax.set_xlabel('Duration')
ax.set_ylabel('Price')
ax.set_zlabel('Synthetic Target (Z)')
ax.set_title('Task 2: Matplotlib 3D Surface Plot (Synthetic Duration-Price Surface)')
fig.colorbar(surf, ax=ax, label='Z Scale')

plt.tight_layout()
plt.show()
```

**TASK7**-Correlation Matrix of Numerical Flight Attributes

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Map string representations of stops to numerical values
stop_mapping = {'zero': 0, 'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5}
df['stops_numeric'] = df['stops'].map(stop_mapping)

# Select the relevant columns, now using the numeric stops column
selected_columns = df[['duration', 'price', 'stops_numeric']]

# Drop rows with NaN values in the selected columns before computing correlation
# This handles cases where 'stops' might have values not in stop_mapping or other NaNs.
selected_columns = selected_columns.dropna()

# Compute the correlation matrix
correlation_matrix = selected_columns.corr()

print("Correlation Matrix of Duration, Price, and Stops:\n")
display(correlation_matrix)

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=.5,
    linecolor='black'
)
plt.title('Correlation Matrix of Duration, Price, and Stops')
plt.show()
```

**TASK10**-Flight Moving Average Price Trend Simulation

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Calculate the 100-flight rolling mean for 'price'
# Use a copy to avoid SettingWithCopyWarning if df was a slice
df_plot = df.copy()
df_plot['rolling_mean_price'] = df_plot['price'].rolling(window=100).mean()

# Plotting the original price and the rolling mean
plt.figure(figsize=(14, 7))
sns.lineplot(data=df_plot, x=df_plot.index, y='price', label='Original Price', alpha=0.7)
sns.lineplot(data=df_plot, x=df_plot.index, y='rolling_mean_price', label='100-Flight Rolling Mean', color='red', linewidth=2)

plt.title('Flight Price Trend with 100-Flight Rolling Mean')
plt.xlabel('Flight Entry Index')
plt.ylabel('Price')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
